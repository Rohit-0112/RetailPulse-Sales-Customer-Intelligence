```python
kpi_summary = {
    "Total Sales": df_clean["Sales"].sum(),
    "Total Profit": df_clean["Profit"].sum(),
    "Total Quantity": df_clean["Quantity"].sum(),
    "Total Orders": df_clean["Order ID"].nunique(),
    "Total Customers": df_clean["Customer ID"].nunique(),
    "Total Products": df_clean["Product ID"].nunique()
}

kpi_summary
```

```python
overall_margin = (
    df_clean["Profit"].sum() /
    df_clean["Sales"].sum()
)

print("Overall Profit Margin:", overall_margin)
print("Overall Profit Margin %:", round(overall_margin * 100, 2), "%")
```

```python
yearly_performance = (
    df_clean
    .groupby("Order Year")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique"),
        Quantity=("Quantity", "sum")
    )
    .reset_index()
)

yearly_performance["Profit Margin %"] = (
    yearly_performance["Profit"] /
    yearly_performance["Sales"] * 100
).round(2)

yearly_performance
```

```python
category_performance = (
    df_clean
    .groupby("Category")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique"),
        Quantity=("Quantity", "sum")
    )
    .reset_index()
)

category_performance["Profit Margin %"] = (
    category_performance["Profit"] /
    category_performance["Sales"] * 100
).round(2)

category_performance.sort_values(
    "Profit",
    ascending=False
)
```

```python
subcategory_performance = (
    df_clean
    .groupby(["Category", "Sub-Category"])
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique"),
        Quantity=("Quantity", "sum")
    )
    .reset_index()
)

subcategory_performance["Profit Margin %"] = (
    subcategory_performance["Profit"] /
    subcategory_performance["Sales"] * 100
).round(2)

subcategory_performance.sort_values(
    "Profit",
    ascending=False
)
```

```python
df_clean["Discount Band"] = pd.cut(
    df_clean["Discount"],
    bins=[-0.01, 0, 0.2, 0.4, 0.6, 0.8],
    labels=[
        "0%",
        "1-20%",
        "21-40%",
        "41-60%",
        "61-80%"
    ]
)

discount_analysis = (
    df_clean
    .groupby("Discount Band", observed=True)
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique"),
        Quantity=("Quantity", "sum")
    )
    .reset_index()
)

discount_analysis["Profit Margin %"] = (
    discount_analysis["Profit"] /
    discount_analysis["Sales"] * 100
).round(2)

discount_analysis
```

```python
discount_subcategory = (
    df_clean
    .groupby(["Sub-Category", "Discount Band"], observed=True)
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique")
    )
    .reset_index()
)

discount_subcategory["Profit Margin %"] = (
    discount_subcategory["Profit"] /
    discount_subcategory["Sales"] * 100
).round(2)

discount_subcategory
```

**Find the products responsible for the largest losses**

```python
product_profit = (
    df_clean
    .groupby(["Product ID", "Product Name"])
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Quantity=("Quantity", "sum"),
        Orders=("Order ID", "nunique"),
        Avg_Discount=("Discount", "mean")
    )
    .reset_index()
)

loss_products = (
    product_profit[product_profit["Profit"] < 0]
    .sort_values("Profit")
    .head(15)
)

loss_products
```

```python
worst_transactions = (
    df_clean[
        [
            "Order ID",
            "Order Date",
            "Customer Name",
            "Product ID",
            "Product Name",
            "Category",
            "Sub-Category",
            "Sales",
            "Quantity",
            "Discount",
            "Profit"
        ]
    ]
    .sort_values("Profit")
    .head(20)
)

worst_transactions
```

```python
discount_impact = discount_analysis.copy()

total_sales = df_clean["Sales"].sum()

discount_impact["Sales Share %"] = (
    discount_impact["Sales"] /
    total_sales * 100
).round(2)

discount_impact["Profit Contribution %"] = (
    discount_impact["Profit"] /
    df_clean["Profit"].sum() * 100
).round(2)

discount_impact
```

```python
customer_performance = (
    df_clean
    .groupby(["Customer ID", "Customer Name", "Segment"])
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique"),
        Quantity=("Quantity", "sum")
    )
    .reset_index()
)

customer_performance["Profit Margin %"] = (
    customer_performance["Profit"] /
    customer_performance["Sales"] * 100
).round(2)

customer_performance.sort_values(
    "Sales",
    ascending=False
).head(20)
```

```python
customer_performance.sort_values(
    "Profit",
    ascending=False
).head(20)
```

```python
segment_performance = (
    df_clean
    .groupby("Segment")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique"),
        Customers=("Customer ID", "nunique"),
        Quantity=("Quantity", "sum")
    )
    .reset_index()
)

segment_performance["Profit Margin %"] = (
    segment_performance["Profit"] /
    segment_performance["Sales"] * 100
).round(2)

segment_performance["Avg Order Value"] = (
    segment_performance["Sales"] /
    segment_performance["Orders"]
).round(2)

segment_performance
```

```python
region_performance = (
    df_clean
    .groupby("Region")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique"),
        Customers=("Customer ID", "nunique"),
        Quantity=("Quantity", "sum")
    )
    .reset_index()
)

region_performance["Profit Margin %"] = (
    region_performance["Profit"] /
    region_performance["Sales"] * 100
).round(2)

region_performance["Avg Order Value"] = (
    region_performance["Sales"] /
    region_performance["Orders"]
).round(2)

region_performance.sort_values(
    "Sales",
    ascending=False
)
```

```python
monthly_performance = (
    df_clean
    .groupby(["Order Year", "Order Month"])
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique")
    )
    .reset_index()
)

monthly_performance["Profit Margin %"] = (
    monthly_performance["Profit"] /
    monthly_performance["Sales"] * 100
).round(2)

monthly_performance
```

```python
shipping_performance = (
    df_clean
    .groupby("Ship Mode")
    .agg(
        Orders=("Order ID", "nunique"),
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Avg_Shipping_Days=("Shipping Days", "mean"),
        Quantity=("Quantity", "sum")
    )
    .reset_index()
)

shipping_performance["Avg Shipping Days"] = (
    shipping_performance["Avg_Shipping_Days"]
    .round(2)
)

shipping_performance["Profit Margin %"] = (
    shipping_performance["Profit"] /
    shipping_performance["Sales"] * 100
).round(2)

shipping_performance.drop(
    columns=["Avg_Shipping_Days"]
)
```
